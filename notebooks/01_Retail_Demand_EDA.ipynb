```python
# Select a manageable subset for EDA

selected_sales = sales_sample[
    (sales_sample["store_id"] == sales_sample["store_id"].iloc[0])
].copy()

print("Selected rows:", selected_sales.shape)
display(selected_sales.head())
```

```python
id_cols = [
    "id",
    "item_id",
    "dept_id",
    "cat_id",
    "store_id",
    "state_id"
]

daily_columns = [
    col for col in selected_sales.columns
    if col.startswith("d_")
]

sales_long = selected_sales.melt(
    id_vars=id_cols,
    value_vars=daily_columns,
    var_name="d",
    value_name="sales"
)

print("Long-format shape:", sales_long.shape)
display(sales_long.head())
```

```python
sales_long = sales_long.merge(
    calendar[["d", "date", "wm_yr_wk", "weekday", "wday", "month", "year"]],
    on="d",
    how="left"
)

sales_long["date"] = pd.to_datetime(sales_long["date"])

display(sales_long.head())
```
